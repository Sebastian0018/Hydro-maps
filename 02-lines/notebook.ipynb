# Map 02 — Lines: how much does the Charles River meander?

**Question:** The Charles River starts at Echo Lake (Hopkinton) and ends at Boston Harbor. How long is its path compared with a straight line between those points?

**Concept: sinuosity** = river length ÷ straight-line distance (source → mouth).
- 1.0 = straight channel · 1.0–1.5 = winding · above 1.5 = meandering
